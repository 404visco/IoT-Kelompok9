```python
# Import library yang dibutuhkan
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
```

```python
#Fungsi Factorial
def factorial(n):
  if n == 0:
    return 1
  else:
    return n * factorial(n-1)

#Fungsi Deret MacLaurin
def deret_maclaurin(fungsi,nilai_asli, x_val, n):
  x = sp.Symbol('x') 
  trunc_error = []
  approx = []
  hasil= 0
  for i in range(n):
    turunan = sp.diff(fungsi,x,i) 
    turunan_x_0= turunan.subs(x,0) 
    suku = turunan_x_0 * x_val**i /factorial(i)
    hasil+= suku
    
    truncation_error = abs(nilai_asli -hasil)
    trunc_error.append(truncation_error)

    approx.append(hasil)
  return trunc_error, approx
```

```python
print('EKSPERIMEN 1 - sin(x))')
x = sp.Symbol('x')
fungsi = sp.sin(x)
nilai_asli = np.sin(0.2)
x_value = 0.2
n = 20

trunc,approx = deret_maclaurin(fungsi, nilai_asli, x_value, n)
for i in range(len(trunc)):
    print(f'--- Orde ke - {i+1} ---')
    print(f'Error = {trunc[i]}\n')
```

```python

orde = np.arange(20)
plt.figure(figsize = (10,10))
plt.plot(orde, approx, marker='o', label='Aproksimasi')
plt.axhline(y=nilai_asli, linestyle='--', label='Nilai Asli')

plt.xlabel('Orde')
plt.ylabel('Nilai')
plt.title('Grafik hubungan orde dan nilai aproksimasi')
plt.legend()
plt.show()
```

```python
print('EKSPERIMEN 2 - 1/(1-2x)')
x =sp.Symbol('x')
fungsi = 1/(1-2*x)
nilai_asli = 1/(1-2*(0.2))
x_value = 0.2
n= 60

trunc,approx = deret_maclaurin(fungsi, nilai_asli, x_value, n)
for i in range(len(approx)):
    print(f'--- Orde ke - {i+1} ---')
    print(f'Truncation Error = {trunc[i]}\n')
```

```python
orde = np.arange(60)

plt.figure(figsize = (19,8))
plt.plot(orde, approx, marker='o', label='Aproksimasi')
plt.axhline(y=nilai_asli, linestyle='--', label='Nilai Asli')

plt.xlabel('Orde')
plt.ylabel('Nilai')
plt.title('Konvergensi Deret Maclaurin')
plt.legend()
plt.show()
```
