```python
# Import library yang dibutuhkan
import numpy as np
import sympy as sp
```

```python
#Fungsi Factorial
def factorial(n):
  if n == 0:
    return 1
  else:
    return n * factorial(n-1)

#Fungsi Deret MacLaurin
def deret_maclaurin(fungsi,nilai_asli, x_val, n):
  x = sp.Symbol('x') 
  trunc_error = []
  roundoff_error = []
  hasil= 0
  for i in range(n):
    turunan = sp.diff(fungsi,x,i) 
    turunan_x_0= turunan.subs(x,0) 
    suku = turunan_x_0 * x_val**i /factorial(i)
    hasil+= suku
    truncation_error = abs(nilai_asli-float(hasil))
    round_error = abs(hasil.evalf(50)-float(hasil))
    trunc_error.append(truncation_error)
    roundoff_error.append(round_error)
  return trunc_error, roundoff_error
```

```python
print('EKSPERIMEN 1 - sin(x))')
x = sp.Symbol('x')
fungsi = sp.tan(x)
nilai_asli = np.tan(0.2)
x_value = 0.2
n = 10

trunc,roundoff = deret_maclaurin(fungsi, nilai_asli, x_value, n)
for i in range(len(trunc)):
    print(f'--- Orde ke - {i+1} ---')
    print(f'Truncation Error = {trunc[i]}')
    print(f'Roundoff Error = {roundoff[i]}\n')
```

```python
print('EKSPERIMEN 2 - 1/(1-2x)')
x =sp.Symbol('x')
fungsi = 1/(1-2*x)
nilai_asli = 1/(1-2*(0.2))
x_value = 0.2

trunc,roundoff = deret_maclaurin(fungsi, nilai_asli, x_value, n)
for i in range(len(roundoff)):
    print(f'--- Orde ke - {i+1} ---')
    print(f'Truncation Error = {trunc[i]}')
    print(f'Roundoff Error = {roundoff[i]}\n')
```
