```python
import sympy as sp
import numpy as np
```

```python
x = sp.Symbol('x')
fx= 2*x*sp.cos(2*x) - (x-2)**2
dfx = 2*sp.cos(2*x) - 4*x*sp.sin(2*x) - 2*(x-2)
```

```python
epsilon = 10e-5
x0 = 3.5
count = 1

X = sp.Symbol('x')
while abs(fx.subs(X, x0))>epsilon:
    f = fx.subs(x, x0)
    df = dfx.subs(x, x0)
    bagi = f/df
    print(count)
    print(f'Fx: {f}')
    print(f'dFx: {df}')
    print(f'f/df : {bagi}')
    print(x0 - bagi)
    x0 = x0 - bagi
    count+=1
    print('\n')
```

```python
x0 = 0
x1 = 1.5708
epsilon = 10e-5
x = sp.Symbol('x')
fx = x - 0.8 - 0.2*sp.sin(x)
count = 1
while abs(fx.subs(x, x1))>epsilon:
    f = fx.subs(x, x1)
    a = x1-x0
    ff = fx.subs(x, x0)
    pemb = f*a
    peny = f-ff
    bagi = pemb/peny
    hasil = x1 - bagi
    print(count)
    print(f'x0: {x0}')
    print(f'x1: {x1}')
    print(f'f0 : {f}')
    print(f'f1 : {ff}')
    print(f'x1-x0: {a}')
    print('\n')
    print(f'Pembilang: {pemb}')
    print(f'Penyebut: {peny}')
    print(f'Bagi: {bagi}')
    print('\n')
    print(f'Hasil: {hasil}')
    fhasil = fx.subs(x, hasil)
    print(f'fhasil: {fhasil}')
    x0 = x1
    x1 = hasil
    count+=1
    print('\n')
```
