```python
import sympy as sp
import numpy as np
```

```python
x = sp.Symbol('x')
fx= 2*x*sp.cos(2*x) - (x-2)**2
dfx = 2*sp.cos(2*x) - 4*x*sp.sin(2*x) - 2*(x-2)
```

```python
epsilon = 10e-5
x0 = 3.5
count = 1

X = sp.Symbol('x')
while abs(fx.subs(X, x0))>epsilon:
    f = fx.subs(x, x0)
    df = dfx.subs(x, x0)
    bagi = f/df
    print(count)
    print(f'Fx: {f}')
    print(f'dFx: {df}')
    print(f'f/df : {bagi}')
    print(x0 - bagi)
    x0 = x0 - bagi
    count+=1
    print('\n')
```
