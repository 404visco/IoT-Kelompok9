```python
import sympy as sp
```

```python
#Biseksi
def biseksi(fungsi,rentang: list, epsilon:float):
    x = sp.Symbol('x')
    selang = rentang
    if fungsi.subs(x, selang[0])*fungsi.subs(x,selang[-1])>0:
        return "Ga ada akar"
    while selang[-1]-selang[0]>epsilon:
        c = selang[0]+(selang[-1]-selang[0])/2
        if fungsi.subs(x,selang[0])*fungsi.subs(x, c)<0:
            selang[-1]= c
        else:
            selang[0] = c
    return selang[0]+(selang[-1]-selang[0])/2
```

```python
x = sp.Symbol('x')
fx = 4-x
epsilon = 0.0001
selang = [-5,2]
```

```python
print(biseksi(fx,selang,epsilon))
```
