```python
import sympy as sp
```

```python
#Biseksi
def biseksi(fungsi,rentang: list, epsilon:float):
    x = sp.Symbol('x')
    a= rentang[0]
    b= rentang[-1]
    if  not fungsi.subs(x, a)*fungsi.subs(x,b)<0:
        return "Ga ada akar"
    while b-a>epsilon:
        c = a + (b-a)/2
        if fungsi.subs(x,a)*fungsi.subs(x, c)<0:
            b= c
        else:
            a = c
    return a+(b-a)/2
```

```python
x = sp.Symbol('x')
fx = 4-x**2
epsilon = 0.0001
selang = [-1,9]
print(biseksi(fx,selang,epsilon))
```

```python
#Regulasi Falsi
def regulasi_falsi(fungsi,rentang,epsilon):
    x = sp.Symbol('x')
    a = rentang[0]
    b = rentang[-1]
    if not fungsi.subs(x,a)*fungsi.subs(x,b)<0:
        return "Ga ada Akar"
    while b-a > epsilon:
        c = b - (fungsi.subs(x, b)*(b-a))/(fungsi.subs(x,b)-fungsi.subs(x,a))
        if abs(fungsi.subs(x,c)) < epsilon:
            break
        elif fungsi.subs(x,a)*fungsi.subs(x,c)<0:
            b = c
        else:
            a = c
    return float(c)
```

```python
x = sp.Symbol('x')
fx = 4-x**2
epsilon = 0.0001
selang = [-1,9]
print(regulasi_falsi(fx,selang,epsilon))
```

```python
#newton-raphson
def newton_raphson(fungsi,tebakan,epsilon):
    x0 = tebakan
    x = sp.Symbol('x')
    fx = fungsi
    dfx = sp.diff(fx)
    while x0>epsilon:
        x0 = x0 - fx.subs(x,x0)/dfx.subs(x,x0)
    return x0
```

```python
x = sp.Symbol('x')
fx = 4-x**2
epsilon = 0.0001
print(newton_raphson(fx, 5, epsilon))
```
