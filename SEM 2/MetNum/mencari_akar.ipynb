## Metode Biseksi (Bisection Method)

Metode Biseksi adalah salah satu **metode numerik** yang digunakan untuk mencari **akar persamaan non-linear**, yaitu nilai ( x ) yang memenuhi:

[
f(x) = 0
]

Metode ini termasuk metode **bracketing (pengurungan akar)** karena akar dicari di dalam suatu interval ([a,b]) yang telah diketahui mengandung akar.

---

### Prinsip Dasar

Metode biseksi bekerja berdasarkan **Teorema Nilai Antara (Intermediate Value Theorem)**, yang menyatakan bahwa:

Jika fungsi ( f(x) ) **kontinu** pada interval ([a,b]) dan

[
f(a)\cdot f(b) < 0
]

maka terdapat minimal satu akar pada interval tersebut.

Artinya:

* jika ( f(a) ) bernilai positif dan ( f(b) ) negatif (atau sebaliknya),
* maka grafik fungsi pasti memotong sumbu-(x) di antara (a) dan (b).

---

### Ide Metode

Interval awal ([a,b]) akan **dibagi dua secara berulang** hingga diperoleh pendekatan akar dengan ketelitian tertentu.

Titik tengah interval dihitung dengan:

$$
c = a+ (b-a)/2
$$

Kemudian dilakukan pengecekan tanda fungsi:

* Jika ( f(a)\cdot f(c) < 0 ) → akar berada di interval ([a,c])
* Jika ( f(c)\cdot f(b) < 0 ) → akar berada di interval ([c,b])

Interval yang tidak mengandung akar akan dibuang, lalu proses diulang.

---

### Langkah-Langkah Algoritma

1. Tentukan fungsi ( f(x) ).
2. Pilih interval awal ([a,b]) sehingga:
   $$
   f(a)\cdot f(b) < 0
   $$
3. Hitung titik tengah:
   $$
   c = a+ (b-a)/2
   $$
4. Evaluasi nilai ( $f(c)$ ).
5. Tentukan interval baru:

   * jika ($f(a)f(c) < 0 $) → ($ b = c $)
   * jika ( $f(c)f(b) < 0 $) → ($ a = c $)
6. Ulangi langkah hingga memenuhi kriteria berhenti.

---

### Kriteria Berhenti

Iterasi dihentikan jika salah satu kondisi terpenuhi:

$$
|b-a| < \varepsilon
$$

atau

$$
|f(c)| \approx 0
$$

di mana ( $\varepsilon$ ) adalah toleransi kesalahan yang diinginkan.

---

### Kelebihan Metode Biseksi

* Mudah dipahami dan diimplementasikan.
* Selalu konvergen jika syarat awal terpenuhi.
* Stabil untuk berbagai fungsi kontinu.

---

### Kekurangan Metode Biseksi

* Konvergensi relatif lambat dibanding metode lain (misalnya Newton-Raphson).
* Membutuhkan interval awal yang mengurung akar.
* Tidak efisien untuk presisi sangat tinggi.

---

### Ilustrasi Konsep

Secara intuitif, metode ini seperti **mencari posisi suatu titik di garis dengan terus membagi dua bagian yang masih mungkin mengandung jawaban**, sehingga rentang pencarian semakin kecil sampai mendekati akar sebenarnya.

---

### Kesimpulan

Metode Biseksi merupakan metode numerik yang sederhana namun sangat andal untuk menentukan akar persamaan non-linear. Dengan prinsip pembagian interval secara berulang, metode ini mampu memberikan pendekatan solusi dengan tingkat kesalahan yang dapat dikontrol melalui parameter toleransi.


```python
#Biseksi
def biseksi(fungsi,rentang:list, epsilon:float):
    x = sp.Symbol('x')
    a= rentang[0]
    b= rentang[-1]
    if  not fungsi.subs(x, a)*fungsi.subs(x,b)<0:
        return "Ga ada akar"
    while abs(b-a)>epsilon:
        c = a + (b-a)/2
        if fungsi.subs(x,a)*fungsi.subs(x, c)<0:
            b= c
        else:
            a = c
    return a+(b-a)/2
```

```python
x = sp.Symbol('x')
fx = 4-x**2
epsilon = 0.0001
selang = [-1,9]
print(biseksi(fx,selang,epsilon))
```

## Metode Regula Falsi (False Position Method)

Metode Regula Falsi adalah metode numerik yang digunakan untuk mencari **akar persamaan non-linear**, yaitu nilai ( x ) yang memenuhi:

$$
f(x)=0
$$

Metode ini termasuk metode **bracketing**, sama seperti metode biseksi, karena akar dicari di dalam suatu interval awal ([a,b]) yang mengurung akar.

Perbedaannya, metode regula falsi **tidak membagi interval menjadi dua**, melainkan menggunakan **garis lurus (interpolasi linear)** untuk memperkirakan posisi akar secara lebih akurat.

---

### Prinsip Dasar

Metode ini juga berdasarkan **Teorema Nilai Antara**, yaitu:

Jika fungsi kontinu pada interval ([a,b]) dan memenuhi

$$
f(a),f(b) < 0
$$

maka terdapat minimal satu akar pada interval tersebut.

Artinya nilai fungsi di ujung interval memiliki tanda yang berbeda sehingga grafik fungsi memotong sumbu-(x).

---

### Ide Metode

Alih-alih mengambil titik tengah seperti pada metode biseksi, metode regula falsi membentuk **garis lurus** yang menghubungkan dua titik:

$$
(a,f(a)) \quad \text{dan} \quad (b,f(b))
$$

Titik potong garis tersebut dengan sumbu-(x) digunakan sebagai pendekatan akar baru.

---

### Penurunan Rumus Titik Pendekatan (c) pada Metode Regula Falsi

Metode Regula Falsi menentukan titik pendekatan akar dengan menggunakan **interpolasi linear**, yaitu membentuk garis lurus yang melalui dua titik pada grafik fungsi:

$$
(a,f(a)) \quad \text{dan} \quad (b,f(b))
$$

Titik potong garis tersebut dengan sumbu-(x) menjadi pendekatan akar yang dinotasikan sebagai (c).

**1. Gradien Garis Penghubung**

Gradien (kemiringan) garis lurus yang melalui dua titik diberikan oleh rumus:

$$
m = \frac{f(b)-f(a)}{b-a}
$$

Gradien ini menyatakan perubahan nilai fungsi terhadap perubahan variabel (x).


**2. Persamaan Garis Lurus**

Persamaan garis dengan gradien (m) yang melalui titik ((a,f(a))) dapat ditulis menggunakan bentuk titik–gradien:

$$
y - f(a) = m(x-a)
$$

Substitusi nilai gradien menghasilkan:

$$
y - f(a) = \frac{f(b)-f(a)}{b-a}(x-a)
$$


**3. Menentukan Titik Potong Sumbu-(x)**

Akar pendekatan terjadi ketika garis memotong sumbu-(x), yaitu saat:

$$
y = 0
$$

Sehingga:

$$
f(a) = \frac{f(b)-f(a)}{b-a}(x-a)
$$

**4. Menyelesaikan terhadap (x)**

Kalikan kedua ruas dengan ((b-a)):

$$

 f(a)(b-a) = (f(b)-f(a))(x-a) 
$$

Bagi kedua ruas dengan ((f(b)-f(a))):

$$
x-a = \frac{-f(a)(b-a)}{f(b)-f(a)}
$$

Sehingga diperoleh:

$$
x = a - \frac{f(a)(b-a)}{f(b)-f(a)}
$$

Nilai (x) inilah yang menjadi titik pendekatan akar, yaitu:

$$
c = a - \frac{f(a)(b-a)}{f(b)-f(a)}
$$


**5. Bentuk Rumus Alternatif**

Rumus di atas dapat ditulis dalam bentuk lain yang ekuivalen:

$$
c = b - \frac{f(b)(b-a)}{f(b)-f(a)}
$$

Bentuk ini sering digunakan dalam implementasi program karena langsung menggunakan titik ujung kanan interval.

---

Langkah-Langkah Algoritma

1. Tentukan fungsi ( $f(x)$ ).
2. Pilih interval awal ([a,b]) sehingga:
   $$
   f(a)f(b) < 0
   $$
3. Hitung titik pendekatan:
   $$
   c = b - \frac{f(b)(b-a)}{f(b)-f(a)}
   $$
4. Hitung nilai ( $f(c) $).
5. Tentukan interval baru:

   * jika ( $f(a)f(c) < 0 $) → akar berada di ([a,c]), maka ($ b=c$ )
   * jika ( $f(c)f(b) < 0 $) → akar berada di ([c,b]), maka ($ a=c $)
6. Ulangi proses hingga memenuhi kriteria berhenti.

---

### Kriteria Berhenti

Iterasi dihentikan apabila salah satu kondisi berikut terpenuhi:

$$
|f(c)| < \varepsilon
$$

atau

$$
|b-a| < \varepsilon
$$

dengan ( \varepsilon ) adalah toleransi kesalahan yang diinginkan.

---

### Hubungan dengan Implementasi Program

Pada implementasi program:

* Interval awal diperiksa terlebih dahulu untuk memastikan akar terkurung:
  $$
  f(a)f(b) < 0
  $$
* Nilai (c) dihitung menggunakan rumus interpolasi linear.
* Interval diperbarui berdasarkan perubahan tanda fungsi, sama seperti metode biseksi.
* Iterasi berhenti ketika nilai fungsi sudah cukup dekat dengan nol atau interval sudah sangat kecil.

---

### Kelebihan Metode Regula Falsi

* Biasanya lebih cepat dibanding metode biseksi.
* Tetap stabil karena akar selalu berada dalam interval.
* Mudah diimplementasikan secara komputasional.

---

### Kekurangan Metode Regula Falsi

* Konvergensi dapat melambat jika salah satu ujung interval jarang berubah.
* Tidak secepat metode Newton-Raphson pada beberapa kasus.
* Tetap membutuhkan interval awal yang mengurung akar.

---

### Intuisi Geometris

Metode ini dapat dibayangkan sebagai proses menarik **garis lurus antara dua titik pada grafik fungsi**, lalu menggunakan titik potong garis tersebut dengan sumbu-(x) sebagai perkiraan akar. Proses ini diulang terus dengan interval yang semakin mendekati akar sebenarnya.

---

### Kesimpulan

Metode Regula Falsi merupakan pengembangan dari metode biseksi dengan memanfaatkan interpolasi linear untuk memperoleh pendekatan akar yang lebih akurat. Metode ini tetap menjaga kestabilan pencarian akar karena selalu mempertahankan interval yang mengurung akar, namun umumnya memberikan konvergensi yang lebih cepat dibanding metode biseksi.


```python
#Regulasi Falsi
def regulasi_falsi(fungsi,rentang:list,epsilon:float):
    x = sp.Symbol('x')
    a = rentang[0]
    b = rentang[-1]
    if not fungsi.subs(x,a)*fungsi.subs(x,b)<0:
        return "Ga ada Akar"
    while b-a > epsilon:
        c = b - (fungsi.subs(x, b)*(b-a))/(fungsi.subs(x,b)-fungsi.subs(x,a))
        if abs(fungsi.subs(x,c)) < epsilon:
            break
        elif fungsi.subs(x,a)*fungsi.subs(x,c)<0:
            b = c
        else:
            a = c
    return float(c)
```

```python
x = sp.Symbol('x')
fx = 4-x**2
epsilon = 0.0001
selang = [-4,1]
print(regulasi_falsi(fx,selang,epsilon))
```

## Metode Newton–Raphson (Newton's Method)

Metode Newton–Raphson adalah metode numerik yang digunakan untuk mencari **akar persamaan non-linear**, yaitu nilai ( x ) yang memenuhi:

$$
f(x)=0
$$

Berbeda dengan metode biseksi dan regula falsi yang menggunakan interval, metode Newton–Raphson termasuk **metode terbuka (open method)** karena hanya membutuhkan **satu tebakan awal** untuk memulai iterasi.

Metode ini terkenal karena memiliki **konvergensi sangat cepat** apabila tebakan awal cukup dekat dengan akar sebenarnya.

---

### Prinsip Dasar

Metode Newton–Raphson menggunakan konsep **garis singgung (tangent line)** pada grafik fungsi.

Ide utamanya adalah:

* ambil suatu titik tebakan awal (x_0),
* buat garis singgung kurva (f(x)) di titik tersebut,
* titik potong garis singgung dengan sumbu-(x) digunakan sebagai pendekatan akar berikutnya.

Proses ini diulang hingga diperoleh nilai yang cukup mendekati akar.

---

### Penurunan Rumus Iterasi

Misalkan titik pada kurva adalah:

$$
(x_0, f(x_0))
$$

Gradien garis singgung pada titik tersebut adalah turunan fungsi:

$$
f'(x_0)
$$

Persamaan garis singgung (bentuk titik–gradien):

$$
y - f(x_0) = f'(x_0)(x - x_0)
$$

Akar pendekatan terjadi saat garis singgung memotong sumbu-(x), yaitu ketika:

$$
y = 0
$$

Sehingga:

$$
-f(x_0) = f'(x_0)(x - x_0)
$$

Menyelesaikan terhadap (x):

$$
x = x_0 - \frac{f(x_0)}{f'(x_0)}
$$

Nilai ini menjadi pendekatan baru yang dinotasikan sebagai:

$$
x_1 = x_0 - \frac{f(x_0)}{f'(x_0)}
$$

Secara umum, rumus iterasi Newton–Raphson adalah:

$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$

---

### Langkah-Langkah Algoritma

1. Tentukan fungsi ( f(x) ).
2. Tentukan tebakan awal ( x_0 ).
3. Hitung turunan fungsi ( f'(x) ).
4. Hitung pendekatan baru:
   [
   x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
   ]
5. Ulangi proses hingga memenuhi kriteria berhenti.

---

### Kriteria Berhenti

Iterasi dihentikan jika perubahan nilai sudah sangat kecil:

$$
|x_{n+1}-x_n| < \varepsilon
$$

atau jumlah iterasi telah mencapai batas maksimum.

Parameter ( \varepsilon ) menyatakan toleransi kesalahan yang diinginkan.

---

### Hubungan dengan Implementasi Program

Pada implementasi program:

* Tebakan awal diberikan sebagai nilai awal iterasi.
* Turunan fungsi didapat dengan fungsi diff dari library sympy.
* Nilai pendekatan baru dihitung menggunakan rumus Newton–Raphson.
* Iterasi berhenti ketika selisih dua pendekatan berturut-turut lebih kecil dari toleransi error.

---

### Kelebihan Metode Newton–Raphson

* Konvergensi sangat cepat (orde kuadratik).
* Membutuhkan iterasi lebih sedikit dibanding metode bracketing.
* Efisien untuk perhitungan presisi tinggi.

---

### Kekurangan Metode Newton–Raphson

* Membutuhkan turunan fungsi.
* Sensitif terhadap tebakan awal.
* Dapat gagal konvergen jika:

  * tebakan awal terlalu jauh dari akar,
  * turunan mendekati nol,
  * fungsi memiliki bentuk yang kompleks.

---

### Intuisi Geometris

Metode ini dapat dibayangkan sebagai proses **melompat dari satu garis singgung ke garis singgung berikutnya**. Setiap garis singgung memberikan estimasi baru yang biasanya langsung mendekati akar sebenarnya dengan sangat cepat.

---

### Kesimpulan

Metode Newton–Raphson merupakan metode pencarian akar yang sangat efisien dengan memanfaatkan informasi turunan fungsi. Dengan pendekatan garis singgung, metode ini mampu mencapai solusi dengan kecepatan konvergensi tinggi, meskipun membutuhkan tebakan awal yang baik agar proses iterasi berhasil.


```python
#newton-raphson
def newton_raphson(fungsi,tebakan:float,epsilon:float,iter_maks = 100):
    x0 = tebakan
    x = sp.Symbol('x')
    fx = fungsi
    dfx = sp.diff(fx,x)
    i = 0
    while i<iter_maks:
        x1 = x0 - fx.subs(x,x0)/dfx.subs(x,x0)
        if abs(x1-x0)<=epsilon:
            return float(x1)
        x0 = x1

        i+=1
    return x1
```

```python
x = sp.Symbol('x')
fx = 4-x**2
epsilon = 0.0001
print(newton_raphson(fx, 5, epsilon))
```

## Metode Secant (Secant Method)

Metode Secant adalah metode numerik untuk mencari **akar persamaan non-linear**, yaitu nilai ( x ) yang memenuhi:

$$
f(x)=0
$$

Metode ini termasuk **metode terbuka (open method)** seperti Newton–Raphson, tetapi **tidak memerlukan turunan fungsi**. Sebagai gantinya, metode secant menggunakan pendekatan kemiringan garis berdasarkan dua titik sebelumnya.

Metode ini dapat dianggap sebagai pengembangan dari Newton–Raphson dengan turunan yang diganti menggunakan pendekatan numerik.

---

### Prinsip Dasar

Alih-alih menggunakan garis singgung seperti Newton–Raphson, metode secant menggunakan **garis secant**, yaitu garis lurus yang melalui dua titik pada kurva:

$$
(x_{n-1}, f(x_{n-1})) \quad \text{dan} \quad (x_n, f(x_n))
$$

Titik potong garis tersebut dengan sumbu-(x) digunakan sebagai pendekatan akar berikutnya.

---

### Pendekatan Gradien

Gradien garis secant diperoleh dari dua titik:

$$
m \approx \frac{f(x_n)-f(x_{n-1})}{x_n-x_{n-1}}
$$

Nilai ini merupakan pendekatan turunan:

$$
f'(x_n) \approx \frac{f(x_n)-f(x_{n-1})}{x_n-x_{n-1}}
$$

---

### Penurunan Rumus Iterasi

Rumus Newton–Raphson adalah:

$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$

Karena turunan tidak digunakan secara langsung, turunan diganti dengan pendekatan gradien secant:

$$
x_{n+1}
= x_n -
\frac{f(x_n)(x_n-x_{n-1})}{f(x_n)-f(x_{n-1})}
$$

Inilah rumus utama metode secant.

---

### Langkah-Langkah Algoritma

1. Tentukan fungsi ( f(x) ).
2. Pilih dua tebakan awal:
   $$
   x_0 \quad \text{dan} \quad x_1
   $$
3. Hitung pendekatan baru:
   $$
   x_{n+1}
   = x_n -
   \frac{f(x_n)(x_n-x_{n-1})}{f(x_n)-f(x_{n-1})}
   $$
4. Geser nilai:
   $$
   x_{n-1} \leftarrow x_n,\quad x_n \leftarrow x_{n+1}
   $$
5. Ulangi hingga memenuhi kriteria berhenti.

---

### Kriteria Berhenti

Iterasi dihentikan apabila:

$$
|x_{n+1}-x_n| < \varepsilon
$$

atau jumlah iterasi mencapai batas maksimum.

Parameter ( \varepsilon ) menyatakan toleransi kesalahan yang diinginkan.

---

### Hubungan dengan Implementasi Program

Pada implementasi program:

* Dua tebakan awal digunakan sebagai dasar pembentukan garis secant.
* Nilai fungsi dihitung pada kedua titik tersebut.
* Pendekatan baru dihitung menggunakan rumus secant.
* Iterasi berhenti jika perubahan nilai sudah lebih kecil dari toleransi error.
* Program juga memeriksa kondisi pembagi nol ketika:
  $$
  f(x_n)-f(x_{n-1}) = 0
  $$
  karena kondisi ini membuat iterasi tidak dapat dilanjutkan.

---

### Kelebihan Metode Secant

* Tidak membutuhkan turunan fungsi.
* Biasanya lebih cepat dibanding metode biseksi dan regula falsi.
* Konvergensi mendekati Newton–Raphson.

---

### Kekurangan Metode Secant

* Tidak selalu stabil.
* Sensitif terhadap pemilihan tebakan awal.
* Dapat gagal konvergen pada fungsi tertentu.
* Tidak menjamin akar selalu berada dalam interval tertentu.

---

### Intuisi Geometris

Metode secant dapat dibayangkan sebagai proses menarik garis lurus yang menghubungkan dua titik terakhir pada grafik fungsi. Titik potong garis tersebut dengan sumbu-(x) menjadi estimasi akar berikutnya. Setiap iterasi menggunakan dua titik terbaru sehingga pendekatan bergerak menuju akar.

---

### Kesimpulan

Metode Secant merupakan metode pencarian akar yang efisien karena tidak memerlukan turunan fungsi namun tetap memiliki kecepatan konvergensi tinggi. Dengan menggunakan pendekatan garis secant secara berulang, metode ini mampu menghasilkan estimasi akar yang semakin akurat hingga mencapai batas toleransi kesalahan yang ditentukan.


```python
def secant(fungsi,tebak0:float,tebak1:float,epsilon:float, iter_maks= 20):
    fx = fungsi
    x1= tebak1
    x0= tebak0
    for i in range(iter_maks):
        fx1 = fx(x1)
        fx0 = fx(x0)

        if fx1-fx0 == 0:
            return "Cari tebakan lain (error: pembagi nol)"
        
        xnew= x1 - fx1*(x1-x0)/(fx1-fx0)
        if abs(xnew - x1) < epsilon:
            return float(xnew)
        x0 = x1
        x1 = xnew
    return float(x1)
```

```python
x = sp.Symbol('x')
f = x*sp.E**(-x) + 1
fx = sp.lambdify(x, f, 'math')
epsilon = 0.0001
print(secant(fx, -1,0, epsilon))
```
