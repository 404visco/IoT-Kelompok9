```python
import sympy as sp
from tabulate import tabulate
```

```python
# Helper untuk menampilkan header kasus dan hasil

# Fungsi untuk mencetak header kasus
def print_header_case(method, function):
    print('=' * 50)
    print(f'METHOD: {method}')
    print(f'Function: {function}')
    print('=' * 50)
    
# Fungsi untuk mencetak tabel iterasi Secant
def print_secant_table(root: float, iteration_history: list[tuple[int, float, float, float, float, float, float]]):
    print(f"Root approximation: {root:.6f}")
    print(f"Total iterations: {len(iteration_history)}")
    
    headers = ["Iterasi", "x_{n-1}", "f(x_{n-1})", "x_n", "f(x_n)", "x_{n+1}", "|x_{n+1} - x_n|"]
    table = tabulate(iteration_history, headers=headers, tablefmt="grid", floatfmt=".6f")
    print(table)

# Fungsi untuk mencetak tabel iterasi Newton-Raphson 
def print_newton_raphson_table(root: float, iteration_history: list[tuple[int, float, float, float, float]]):
    print(f"Root approximation: {root:.6f}")
    print(f"Total Iterations: {len(iteration_history)}")
    
    headers = ["Iterasi", "x_n", "f(x_n)", "f'(x_n)", "Error"]
    table = tabulate(iteration_history, headers=headers, tablefmt="grid", floatfmt=".6f")
    print(table)

# Fungsi untuk mencetak tabel iterasi Regula Falsi
def print_regula_falsi_table(root: float, iteration_data: list[tuple[int, float, float, float, float, float, float, float]]):
    print(f"Root approximation: {root:.6f}")
    print(f"Total iterations: {len(iteration_data)}")
    
    headers = ["Iteration", "a", "b", "c", "f(a)", "f(b)", "f(c)", "Error Estimate\n(|c_n - c_{n-1}|)"]
    print(tabulate(iteration_data, headers=headers, tablefmt="grid", floatfmt=".6f")) 

# Fungsi untuk mencetak tabel iterasi Bisection
def print_bisection_table(root: float, iteration_data: list[tuple[int, float, float, float, float, float]]):
    print(f"Root approximation: {root:.6f}")
    print(f"Total iterations: {len(iteration_data)}")
    
    headers = ["Iteration", "a", "b", "c", "f(c)", "Error Estimate\n(Interval Width)"]
    print(tabulate(iteration_data, headers=headers, tablefmt="grid", floatfmt=".6f"))

# Fungsi untuk memperkirakan jumlah iterasi yang dibutuhkan untuk konvergensi
# Adaptasi dari rumus iterasi bisection: n >= log2((b - a) / tolerance) 
def iteration_estimate(a: float, b: float, tolerance: float) -> int:
    return int(sp.ceiling(sp.log((b - a) / tolerance, 2)))

# Fungsi untuk mencetak hasil perkiraan iterasi dan membandingkannya dengan iterasi aktual
def print_verify_iteration_estimate(a: float, b: float, tolerance: float, actual_iterations: int):
    estimated = iteration_estimate(a, b, tolerance)
    print(f"Estimated iterations needed for convergence: {estimated}")
    print(f"Actual iterations taken: {actual_iterations}")
    if actual_iterations <= estimated:
        print("The actual iterations are within the estimated bound.")
    else:
        print("The actual iterations exceed the estimated bound. Consider reviewing the implementation or parameters.")
```

## Metode Biseksi (Bisection Method)

Metode Biseksi adalah salah satu **metode numerik** yang digunakan untuk mencari **akar persamaan non-linear**, yaitu nilai $x$ yang memenuhi:

$$
f(x) = 0
$$

Metode ini termasuk metode **bracketing (pengurungan akar)** karena akar dicari di dalam suatu interval $[a,b]$ yang telah diketahui mengandung akar.

---

## Prinsip Dasar

Metode biseksi bekerja berdasarkan **Teorema Nilai Antara (Intermediate Value Theorem)**, yang menyatakan bahwa:

Jika fungsi $f(x)$ kontinu pada interval $[a,b]$ dan

$$
f(a)\cdot f(b) < 0
$$

maka terdapat minimal satu akar pada interval tersebut.

Artinya:

- Jika $f(a)$ bernilai positif dan $f(b)$ negatif (atau sebaliknya),
- Maka grafik fungsi pasti memotong sumbu-$x$ di antara $a$ dan $b$.

Metode ini mensyaratkan fungsi kontinu pada interval yang dipilih, karena tanpa kontinuitas Teorema Nilai Antara tidak dapat diterapkan.

---

## Ide Metode

Interval awal $[a,b]$ akan dibagi dua secara berulang hingga diperoleh pendekatan akar dengan ketelitian tertentu.

Titik tengah interval dihitung dengan:

$$
c = a + \frac{b-a}{2}
$$

Kemudian dilakukan pengecekan tanda fungsi:

- Jika $f(a)\cdot f(c) < 0$, maka akar berada di interval $[a,c]$
- Jika $f(c)\cdot f(b) < 0$, maka akar berada di interval $[c,b]$

Interval yang tidak mengandung akar akan dibuang, lalu proses diulang.

---

## Langkah-Langkah Algoritma

1. Tentukan fungsi $f(x)$.
2. Pilih interval awal $[a,b]$ sehingga:
   $$
   f(a)\cdot f(b) < 0
   $$
3. Hitung titik tengah:
   $$
   c = a + \frac{b-a}{2}
   $$
4. Evaluasi nilai $f(c)$.
5. Tentukan interval baru:
   - Jika $f(a)\cdot f(c) < 0$, maka $b = c$
   - Jika $f(c)\cdot f(b) < 0$, maka $a = c$
6. Ulangi langkah hingga memenuhi kriteria berhenti.

---

## Kriteria Berhenti

Iterasi dihentikan jika salah satu kondisi terpenuhi:

$$
|b-a| < \varepsilon
$$

atau

$$
|f(c)| \approx 0
$$

di mana $\varepsilon$ adalah toleransi kesalahan yang diinginkan.

---

## Analisis Galat (Error Analysis)

Pada metode biseksi, galat maksimum setelah iterasi ke-$n$ dibatasi oleh:

$$
|x_n - r| \le \frac{b-a}{2^n}
$$

di mana:
- $r$ adalah akar sebenarnya,
- $x_n$ adalah aproksimasi pada iterasi ke-$n$.

Dari persamaan tersebut dapat ditentukan estimasi jumlah iterasi minimum untuk mencapai toleransi $\varepsilon$:

$$
n \ge \log_2 \left( \frac{b-a}{\varepsilon} \right)
$$

Rumus ini memberikan batas teoritis jumlah iterasi yang dibutuhkan untuk mencapai tingkat ketelitian tertentu.

---

## Tingkat Konvergensi

Metode biseksi memiliki **konvergensi linear**, yang berarti galat berkurang secara proporsional terhadap iterasi sebelumnya. Secara khusus, panjang interval akan berkurang setengah pada setiap iterasi.

Hal ini membuat metode ini stabil dan dapat diprediksi, tetapi relatif lebih lambat dibanding metode dengan konvergensi kuadratik seperti Newton-Raphson.

---

## Kelebihan Metode Biseksi

- Mudah dipahami dan diimplementasikan.
- Selalu konvergen jika syarat awal terpenuhi.
- Stabil untuk berbagai fungsi kontinu.

---

## Kekurangan Metode Biseksi

- Konvergensi relatif lambat dibanding metode lain (misalnya Newton-Raphson).
- Membutuhkan interval awal yang mengurung akar.
- Tidak efisien untuk presisi sangat tinggi.

---

## Ilustrasi Konsep

Secara intuitif, metode ini seperti mencari posisi suatu titik di garis dengan terus membagi dua bagian yang masih mungkin mengandung jawaban, sehingga rentang pencarian semakin kecil sampai mendekati akar sebenarnya.

---

## Kesimpulan

Metode Biseksi merupakan metode numerik yang sederhana namun sangat andal untuk menentukan akar persamaan non-linear. Dengan prinsip pembagian interval secara berulang, metode ini mampu memberikan pendekatan solusi dengan tingkat kesalahan yang dapat dikontrol melalui parameter toleransi. Selain itu, metode ini memiliki batas galat yang jelas dan jumlah iterasi yang dapat diperkirakan secara teoritis.

```python
# Fungsi utama untuk metode bisection
def bisection(function: sp.Expr, x_sym: sp.Symbol, bracket_range: tuple[float, float], tolerance: float, max_iterations: int) -> tuple[float, list[tuple], int]:
    """
    Metode Biseksi untuk mencari akar f(x) = 0
    
    Parameters:
    - function: Fungsi yang ingin dicari akarnya, dalam bentuk ekspresi sympy
    - x_sym: Simbol variabel dalam fungsi, sebagai objek sympy.Symbol
    - bracket_range: Tuple (a, b) yang merupakan interval awal dimana akar diperkirakan berada, dengan syarat f(a) dan f(b) memiliki tanda yang berbeda
    - tolerance: Batas toleransi untuk konvergensi, baik dalam nilai fungsi maupun lebar interval
    - max_iterations: Batas maksimum iterasi untuk mencegah loop tak berujung jika konvergensi tidak tercapai
    
    Returns:
    - root: Perkiraan akar yang ditemukan
    - history: Daftar tuple yang berisi data iterasi (iteration, a, b, c, f(c), error_estimate)
    - iteration: Jumlah iterasi yang dilakukan
    """
    
    # Validasi bahwa f(a) dan f(b) memiliki tanda yang berbeda
    a, b = bracket_range
    f = sp.lambdify(x_sym, function, 'numpy')
    f_a = f(a)
    f_b = f(b)
    
    if f_a * f_b >= 0:
        raise ValueError("No root exists in the given interval. The function must have different signs at the endpoints a and b.")
    
    # Inisialisasi variabel untuk iterasi
    history = []
    iteration = 0
    
    # Iterasi metode bisection
    for i in range(max_iterations):
        # Hitung titik tengah dan nilai fungsi di titik tengah
        c = (a + b) / 2
        
        # Evaluasi fungsi di titik tengah
        f_c = f(c)
        
        # Hitung lebar interval sebagai perkiraan error 
        interval_width = (b - a) / 2
        
        # Simpan data iterasi ke history
        history.append((i+1, a, b, c, f_c, interval_width))
        
        # Increment iterasi
        iteration += 1
        
        # Periksa kondisi konvergensi: jika f(c) mendekati nol atau lebar interval sudah cukup kecil
        # Kondisi ini memastikan bahwa kita berhenti jika kita sudah cukup dekat dengan akar atau jika interval sudah cukup kecil untuk dianggap sebagai solusi yang valid
        if abs(f_c) < tolerance or abs(interval_width) < tolerance:
            return c, history, iteration
        
        # Tentukan subinterval berikutnya berdasarkan tanda f(c)
        if f_a * f_c < 0:
            b = c
            f_b = f_c
        else:
            a = c
            f_a = f_c

    # Jika kita mencapai batas iterasi maksimum tanpa konvergensi, kita bisa memberikan peringatan atau mengembalikan hasil terbaik yang ditemukan sejauh ini
    print("Warning: Maximum iterations reached without convergence.")            
    return c, history, iteration
```

```python
# Implementasi Kasus
# Kasus 1: Mencari akar dari fungsi f(x) = xe^-x + 1 dengan interval [-1, 0]
# e : euler's number
# ============================================================
METHOD = "Bisection"

X_SYM = sp.Symbol('x')
FUNC = X_SYM * sp.E**(-X_SYM) + 1
BRACKET_RANGE = (-1, 0)

EPSILON_TOLERANCE = 10**(-3)
MAX_ITERATIONS = 20

# ===========================================================
print_header_case(METHOD, FUNC)
# ===========================================================

# Menjalankan metode bisection
# ===========================================================
root, iteration_history, iteration_count = bisection(FUNC, X_SYM, BRACKET_RANGE, EPSILON_TOLERANCE, MAX_ITERATIONS)
print_bisection_table(root, iteration_history)
# ===========================================================

# Verifikasi jumlah iterasi yang sebenarnya dengan estimasi
# ===========================================================
print_verify_iteration_estimate(BRACKET_RANGE[0], BRACKET_RANGE[1], EPSILON_TOLERANCE, iteration_count)
```

## Metode Regula Falsi (False Position Method)

Metode Regula Falsi adalah metode numerik yang digunakan untuk mencari **akar persamaan non-linear**, yaitu nilai ( x ) yang memenuhi:

$$
f(x)=0
$$

Metode ini termasuk metode **bracketing**, sama seperti metode biseksi, karena akar dicari di dalam suatu interval awal ([a,b]) yang mengurung akar.

Perbedaannya, metode regula falsi **tidak membagi interval menjadi dua**, melainkan menggunakan **garis lurus (interpolasi linear)** untuk memperkirakan posisi akar secara lebih akurat.

---

### Prinsip Dasar

Metode ini juga berdasarkan **Teorema Nilai Antara**, yaitu:

Jika fungsi kontinu pada interval ([a,b]) dan memenuhi

$$
f(a),f(b) < 0
$$

maka terdapat minimal satu akar pada interval tersebut.

Artinya nilai fungsi di ujung interval memiliki tanda yang berbeda sehingga grafik fungsi memotong sumbu-(x).

---

### Ide Metode

Alih-alih mengambil titik tengah seperti pada metode biseksi, metode regula falsi membentuk **garis lurus** yang menghubungkan dua titik:

$$
(a,f(a)) \quad \text{dan} \quad (b,f(b))
$$

Titik potong garis tersebut dengan sumbu-(x) digunakan sebagai pendekatan akar baru.

---

### Penurunan Rumus Titik Pendekatan (c) pada Metode Regula Falsi

Metode Regula Falsi menentukan titik pendekatan akar dengan menggunakan **interpolasi linear**, yaitu membentuk garis lurus yang melalui dua titik pada grafik fungsi:

$$
(a,f(a)) \quad \text{dan} \quad (b,f(b))
$$

Titik potong garis tersebut dengan sumbu-(x) menjadi pendekatan akar yang dinotasikan sebagai (c).

**1. Gradien Garis Penghubung**

Gradien (kemiringan) garis lurus yang melalui dua titik diberikan oleh rumus:

$$
m = \frac{f(b)-f(a)}{b-a}
$$

Gradien ini menyatakan perubahan nilai fungsi terhadap perubahan variabel (x).


**2. Persamaan Garis Lurus**

Persamaan garis dengan gradien (m) yang melalui titik ((a,f(a))) dapat ditulis menggunakan bentuk titik–gradien:

$$
y - f(a) = m(x-a)
$$

Substitusi nilai gradien menghasilkan:

$$
y - f(a) = \frac{f(b)-f(a)}{b-a}(x-a)
$$


**3. Menentukan Titik Potong Sumbu-(x)**

Akar pendekatan terjadi ketika garis memotong sumbu-(x), yaitu saat:

$$
y = 0
$$

Sehingga:

$$
f(a) = \frac{f(b)-f(a)}{b-a}(x-a)
$$

**4. Menyelesaikan terhadap (x)**

Kalikan kedua ruas dengan ((b-a)):

$$

 f(a)(b-a) = (f(b)-f(a))(x-a) 
$$

Bagi kedua ruas dengan ((f(b)-f(a))):

$$
x-a = \frac{-f(a)(b-a)}{f(b)-f(a)}
$$

Sehingga diperoleh:

$$
x = a - \frac{f(a)(b-a)}{f(b)-f(a)}
$$

Nilai (x) inilah yang menjadi titik pendekatan akar, yaitu:

$$
c = a - \frac{f(a)(b-a)}{f(b)-f(a)}
$$


**5. Bentuk Rumus Alternatif**

Rumus di atas dapat ditulis dalam bentuk lain yang ekuivalen:

$$
c = b - \frac{f(b)(b-a)}{f(b)-f(a)}
$$

Bentuk ini sering digunakan dalam implementasi program karena langsung menggunakan titik ujung kanan interval.

---

Langkah-Langkah Algoritma

1. Tentukan fungsi ( $f(x)$ ).
2. Pilih interval awal ([a,b]) sehingga:
   $$
   f(a)f(b) < 0
   $$
3. Hitung titik pendekatan:
   $$
   c = b - \frac{f(b)(b-a)}{f(b)-f(a)}
   $$
4. Hitung nilai ( $f(c) $).
5. Tentukan interval baru:

   * jika ( $f(a)f(c) < 0 $) → akar berada di ([a,c]), maka ($ b=c$ )
   * jika ( $f(c)f(b) < 0 $) → akar berada di ([c,b]), maka ($ a=c $)
6. Ulangi proses hingga memenuhi kriteria berhenti.

---

### Kriteria Berhenti

Iterasi dihentikan apabila salah satu kondisi berikut terpenuhi:

$$
|f(c)| < \varepsilon
$$

atau

$$
|b-a| < \varepsilon
$$

dengan ( \varepsilon ) adalah toleransi kesalahan yang diinginkan.

---

### Hubungan dengan Implementasi Program

Pada implementasi program:

* Interval awal diperiksa terlebih dahulu untuk memastikan akar terkurung:
  $$
  f(a)f(b) < 0
  $$
* Nilai (c) dihitung menggunakan rumus interpolasi linear.
* Interval diperbarui berdasarkan perubahan tanda fungsi, sama seperti metode biseksi.
* Iterasi berhenti ketika nilai fungsi sudah cukup dekat dengan nol atau interval sudah sangat kecil.

---

### Kelebihan Metode Regula Falsi

* Biasanya lebih cepat dibanding metode biseksi.
* Tetap stabil karena akar selalu berada dalam interval.
* Mudah diimplementasikan secara komputasional.

---

### Kekurangan Metode Regula Falsi

* Konvergensi dapat melambat jika salah satu ujung interval jarang berubah.
* Tidak secepat metode Newton-Raphson pada beberapa kasus.
* Tetap membutuhkan interval awal yang mengurung akar.

---

### Intuisi Geometris

Metode ini dapat dibayangkan sebagai proses menarik **garis lurus antara dua titik pada grafik fungsi**, lalu menggunakan titik potong garis tersebut dengan sumbu-(x) sebagai perkiraan akar. Proses ini diulang terus dengan interval yang semakin mendekati akar sebenarnya.

---

### Kesimpulan

Metode Regula Falsi merupakan pengembangan dari metode biseksi dengan memanfaatkan interpolasi linear untuk memperoleh pendekatan akar yang lebih akurat. Metode ini tetap menjaga kestabilan pencarian akar karena selalu mempertahankan interval yang mengurung akar, namun umumnya memberikan konvergensi yang lebih cepat dibanding metode biseksi.


```python
# Fungsi utama untuk metode Regula Falsi
def regula_falsi(function: sp.Expr, x_sym: sp.Symbol, bracket_range: tuple[float, float], tolerance: float, max_iterations: int) -> tuple[float, list[tuple], int]:
    """
    Metode Regula Falsi untuk mencari akar f(x) = 0
    
    Parameters:
    - function: Fungsi yang ingin dicari akarnya, dalam bentuk ekspresi sympy
    - x_sym: Simbol variabel dalam fungsi, sebagai objek sympy.Symbol
    - bracket_range: Tuple (a, b) yang merupakan interval awal dimana akar diperkirakan berada, dengan syarat f(a) dan f(b) memiliki tanda yang berbeda
    - tolerance: Batas toleransi untuk konvergensi, baik dalam nilai fungsi maupun lebar interval
    - max_iterations: Batas maksimum iterasi untuk mencegah loop tak berujung jika konvergensi tidak tercapai
    
    Returns:
    - root: Perkiraan akar yang ditemukan
    - history: Daftar tuple yang berisi data iterasi (iteration, a, b, c, f(a), f(b), f(c), error_estimate)
    - iteration: Jumlah iterasi yang dilakukan
    """
    
    # Validasi bahwa f(a) dan f(b) memiliki tanda yang berbeda
    a, b = bracket_range
    f = sp.lambdify(x_sym, function, 'numpy')
    f_a = f(a)
    f_b = f(b)
    
    if f_a * f_b >= 0:
        raise ValueError("No root exists in the given interval. The function must have different signs at the endpoints a and b.")
    
    # Inisialisasi variabel untuk iterasi
    history = []
    iteration = 0
    
    # Untuk menyimpan nilai c dari iterasi sebelumnya, digunakan untuk menghitung error
    c_prev = None  
    
    # Iterasi metode Regula Falsi
    for i in range(max_iterations):
        # Hitung titik pendekatan c 
        c = b - (f_b * (b - a)) / (f_b - f_a)
        
        # Evaluasi f(c) dan hitung lebar interval
        f_c = f(c)
        
        # Hitung error |c_n - c_{n-1}|
        if c_prev is None:
            error = None
        else:
            error = abs(c - c_prev)
                
        # Simpan data iterasi ke dalam history
        history.append((i+1, a, b, c, f_a, f_b, f_c, error))
        
        # Increment jumlah iterasi
        iteration += 1
        
        # Periksa kondisi konvergensi: apakah f(c) cukup dekat dengan nol atau interval sudah cukup kecil
        if abs(f_c) < tolerance or (error is not None and abs(error) < tolerance):
            return c, history, iteration
        
        # Tentukan subinterval berikutnya berdasarkan tanda f(c)
        if f_a * f_c < 0:
            b = c
            f_b = f_c
        else:
            a = c
            f_a = f_c
            
        # Simpan nilai c saat ini sebagai c_prev untuk iterasi berikutnya
        c_prev = c
            
    # Jika kita mencapai batas iterasi maksimum tanpa konvergensi, kita bisa memberikan peringatan atau mengembalikan hasil terbaik yang ditemukan sejauh ini
    print("Warning: Maximum iterations reached without convergence.")            
    return c, history, iteration
```

```python
# Implementasi Kasus
# Kasus 2: Mencari akar dari fungsi f(x) = xe^-x + 1 dengan interval [-1, 0] menggunakan metode Regula Falsi
# e : euler's number
# ============================================================
METHOD = "Regula Falsi"

X_SYM = sp.Symbol('x')
FUNC = X_SYM * sp.E**(-X_SYM) + 1
BRACKET_RANGE = (-1, 0)

EPSILON_TOLERANCE = 10**(-3)
MAX_ITERATIONS = 20

# ===========================================================
print_header_case(METHOD, FUNC)
# ===========================================================

# Menjalankan metode regula falsi
# ===========================================================
root, iteration_history, iteration_count = regula_falsi(FUNC, X_SYM, BRACKET_RANGE, EPSILON_TOLERANCE, MAX_ITERATIONS)
print_regula_falsi_table(root, iteration_history)
# ===========================================================
```

## Metode Newton–Raphson (Newton's Method)

Metode Newton–Raphson adalah metode numerik yang digunakan untuk mencari **akar persamaan non-linear**, yaitu nilai $x$ yang memenuhi:

$$
f(x) = 0
$$

Berbeda dengan metode biseksi dan regula falsi yang menggunakan interval, metode Newton–Raphson termasuk **metode terbuka (open method)** karena hanya membutuhkan **satu tebakan awal** untuk memulai iterasi.

Metode ini terkenal karena memiliki **konvergensi sangat cepat** apabila tebakan awal cukup dekat dengan akar sebenarnya.

---

### Prinsip Dasar

Metode Newton–Raphson menggunakan konsep **garis singgung (tangent line)** pada grafik fungsi.

Ide utamanya adalah:

- Ambil suatu titik tebakan awal $x_0$.
- Buat garis singgung kurva $f(x)$ di titik tersebut.
- Titik potong garis singgung dengan sumbu-$x$ digunakan sebagai pendekatan akar berikutnya.

Proses ini diulang hingga diperoleh nilai yang cukup mendekati akar.

---

### Penurunan Rumus Iterasi

Misalkan titik pada kurva adalah:

$$
(x_0, f(x_0))
$$

Gradien garis singgung pada titik tersebut adalah turunan fungsi:

$$
f'(x_0)
$$

Persamaan garis singgung (bentuk titik–gradien):

$$
y - f(x_0) = f'(x_0)(x - x_0)
$$

Akar pendekatan terjadi saat garis singgung memotong sumbu-$x$, yaitu ketika:

$$
y = 0
$$

Sehingga:

$$
-f(x_0) = f'(x_0)(x - x_0)
$$

Menyelesaikan terhadap $x$ diperoleh:

$$
x = x_0 - \frac{f(x_0)}{f'(x_0)}
$$

Nilai ini menjadi pendekatan baru yang dinotasikan sebagai:

$$
x_1 = x_0 - \frac{f(x_0)}{f'(x_0)}
$$

Secara umum, rumus iterasi Newton–Raphson adalah:

$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$

---

### Langkah-Langkah Algoritma

1. Tentukan fungsi $f(x)$.
2. Tentukan tebakan awal $x_0$.
3. Hitung turunan fungsi $f'(x)$.
4. Hitung pendekatan baru menggunakan rumus:

   $$
   x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
   $$

5. Ulangi proses hingga memenuhi kriteria berhenti.

---

### Kriteria Berhenti

Iterasi dihentikan jika perubahan nilai sudah sangat kecil:

$$
|x_{n+1} - x_n| < \varepsilon
$$

atau jumlah iterasi telah mencapai batas maksimum.

Parameter $\varepsilon$ menyatakan toleransi kesalahan yang diinginkan.

---

### Hubungan dengan Implementasi Program

Pada implementasi program:

- Tebakan awal diberikan sebagai nilai awal iterasi.
- Turunan fungsi diperoleh menggunakan fungsi `diff` dari library `sympy`.
- Nilai pendekatan baru dihitung menggunakan rumus Newton–Raphson.
- Iterasi berhenti ketika selisih dua pendekatan berturut-turut lebih kecil dari toleransi error.

---

### Kelebihan Metode Newton–Raphson

- Konvergensi sangat cepat (orde kuadratik).
- Membutuhkan iterasi lebih sedikit dibanding metode bracketing.
- Efisien untuk perhitungan presisi tinggi.

---

### Kekurangan Metode Newton–Raphson

- Membutuhkan turunan fungsi.
- Sensitif terhadap tebakan awal.
- Dapat gagal konvergen jika:
  - Tebakan awal terlalu jauh dari akar.
  - Turunan mendekati nol.
  - Fungsi memiliki bentuk yang kompleks.

---

### Intuisi Geometris

Metode ini dapat dibayangkan sebagai proses melompat dari satu garis singgung ke garis singgung berikutnya. Setiap garis singgung memberikan estimasi baru yang biasanya langsung mendekati akar sebenarnya dengan sangat cepat.

---

### Kesimpulan

Metode Newton–Raphson merupakan metode pencarian akar yang sangat efisien dengan memanfaatkan informasi turunan fungsi. Dengan pendekatan garis singgung, metode ini mampu mencapai solusi dengan kecepatan konvergensi tinggi, meskipun membutuhkan tebakan awal yang baik agar proses iterasi berhasil.

```python
def newton_raphson(function: sp.Expr, x_sym: sp.Symbol, x0: float, tolerance: float, max_iterations: int) -> tuple[float, list[tuple], int]:
    """
    Menghitung akar dari fungsi non-linear menggunakan metode Newton-Raphson.
    
    Parameters:
    - function: Fungsi yang ingin dicari akarnya, dalam bentuk ekspresi sympy
    - x_sym: Simbol variabel dalam fungsi, sebagai objek sympy.Symbol
    - x0: Nilai awal (initial guess) untuk iterasi Newton-Raphson
    - tolerance: Batas toleransi untuk konvergensi, baik dalam nilai fungsi maupun perubahan nilai x
    - max_iterations: Batas maksimum iterasi untuk mencegah loop tak berujung jika konvergensi tidak tercapai
    
    Returns:
    - root: Perkiraan akar yang ditemukan
    - history: Daftar tuple yang berisi data iterasi (iteration, x_n, f(x_n), f'(x_n), error)
    - iteration: Jumlah iterasi yang dilakukan
    """
    
    # Membuat turunan pertama dari fungsi
    f_prime_expr = sp.diff(function, x_sym)
    
    # Membuat fungsi numerik dari ekspresi simbolik
    f = sp.lambdify(x_sym, function, 'numpy')
    f_prime = sp.lambdify(x_sym, f_prime_expr, 'numpy')
    
    # Inisialisasi variabel untuk iterasi
    history = []
    iteration = 0
    
    # Inisialisasi nilai awal
    x_current = x0
    
    # Iterasi Newton-Raphson
    for i in range(max_iterations):
        # Hitung nilai fungsi dan turunannya di titik saat ini
        f_x = f(x_current)
        f_prime_x = f_prime(x_current)
        
        # Cek apakah turunan nol untuk menghindari pembagian dengan nol
        # Threshold kecil untuk menghindari pembagian dengan nol
        if abs(f_prime_x) < 1e-12: 
            print("Derivative is zero. No solution found.")
            return None
        
        # Hitung nilai x berikutnya menggunakan rumus Newton-Raphson
        x_next = x_current - (f_x / f_prime_x)
        
        # Hitung error sebagai selisih absolut antara x berikutnya dan x saat ini
        error = abs(x_next - x_current)
        
        # Simpan data iterasi ke dalam history
        history.append((i+1, x_current, f_x, f_prime_x, error))
        
        # Increment iterasi
        iteration += 1
        
        # Periksa kriteria konvergensi: nilai fungsi mendekati nol atau error mendekati nol
        if abs(f_x) < tolerance or abs(error) < tolerance:
            return x_next, history, iteration
        
        # Update nilai x saat ini untuk iterasi berikutnya
        x_current = x_next
    
    # Jika kita mencapai batas iterasi maksimum tanpa konvergensi, kita bisa memberikan peringatan atau mengembalikan hasil terbaik yang ditemukan sejauh ini
    print("Maximum iterations reached without convergence.")
    return x_current, history, iteration
```

## Metode Secant (Secant Method)

Metode Secant adalah metode numerik untuk mencari **akar persamaan non-linear**, yaitu nilai ( x ) yang memenuhi:

$$
f(x)=0
$$

Metode ini termasuk **metode terbuka (open method)** seperti Newton–Raphson, tetapi **tidak memerlukan turunan fungsi**. Sebagai gantinya, metode secant menggunakan pendekatan kemiringan garis berdasarkan dua titik sebelumnya.

Metode ini dapat dianggap sebagai pengembangan dari Newton–Raphson dengan turunan yang diganti menggunakan pendekatan numerik.

---

### Prinsip Dasar

Alih-alih menggunakan garis singgung seperti Newton–Raphson, metode secant menggunakan **garis secant**, yaitu garis lurus yang melalui dua titik pada kurva:

$$
(x_{n-1}, f(x_{n-1})) \quad \text{dan} \quad (x_n, f(x_n))
$$

Titik potong garis tersebut dengan sumbu-(x) digunakan sebagai pendekatan akar berikutnya.

---

### Pendekatan Gradien

Gradien garis secant diperoleh dari dua titik:

$$
m \approx \frac{f(x_n)-f(x_{n-1})}{x_n-x_{n-1}}
$$

Nilai ini merupakan pendekatan turunan:

$$
f'(x_n) \approx \frac{f(x_n)-f(x_{n-1})}{x_n-x_{n-1}}
$$

---

### Penurunan Rumus Iterasi

Rumus Newton–Raphson adalah:

$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$

Karena turunan tidak digunakan secara langsung, turunan diganti dengan pendekatan gradien secant:

$$
x_{n+1}
= x_n -
\frac{f(x_n)(x_n-x_{n-1})}{f(x_n)-f(x_{n-1})}
$$

Inilah rumus utama metode secant.

---

### Langkah-Langkah Algoritma

1. Tentukan fungsi ( f(x) ).
2. Pilih dua tebakan awal:
   $$
   x_0 \quad \text{dan} \quad x_1
   $$
3. Hitung pendekatan baru:
   $$
   x_{n+1}
   = x_n -
   \frac{f(x_n)(x_n-x_{n-1})}{f(x_n)-f(x_{n-1})}
   $$
4. Geser nilai:
   $$
   x_{n-1} \leftarrow x_n,\quad x_n \leftarrow x_{n+1}
   $$
5. Ulangi hingga memenuhi kriteria berhenti.

---

### Kriteria Berhenti

Iterasi dihentikan apabila:

$$
|x_{n+1}-x_n| < \varepsilon
$$

atau jumlah iterasi mencapai batas maksimum.

Parameter ( $\varepsilon$ ) menyatakan toleransi kesalahan yang diinginkan.

---

### Hubungan dengan Implementasi Program

Pada implementasi program:

* Dua tebakan awal digunakan sebagai dasar pembentukan garis secant.
* Nilai fungsi dihitung pada kedua titik tersebut.
* Pendekatan baru dihitung menggunakan rumus secant.
* Iterasi berhenti jika perubahan nilai sudah lebih kecil dari toleransi error.
* Program juga memeriksa kondisi pembagi nol ketika:
  $$
  f(x_n)-f(x_{n-1}) = 0
  $$
  karena kondisi ini membuat iterasi tidak dapat dilanjutkan.

---

### Kelebihan Metode Secant

* Tidak membutuhkan turunan fungsi.
* Biasanya lebih cepat dibanding metode biseksi dan regula falsi.
* Konvergensi mendekati Newton–Raphson.

---

### Kekurangan Metode Secant

* Tidak selalu stabil.
* Sensitif terhadap pemilihan tebakan awal.
* Dapat gagal konvergen pada fungsi tertentu.
* Tidak menjamin akar selalu berada dalam interval tertentu.

---

### Intuisi Geometris

Metode secant dapat dibayangkan sebagai proses menarik garis lurus yang menghubungkan dua titik terakhir pada grafik fungsi. Titik potong garis tersebut dengan sumbu-(x) menjadi estimasi akar berikutnya. Setiap iterasi menggunakan dua titik terbaru sehingga pendekatan bergerak menuju akar.

---

### Kesimpulan

Metode Secant merupakan metode pencarian akar yang efisien karena tidak memerlukan turunan fungsi namun tetap memiliki kecepatan konvergensi tinggi. Dengan menggunakan pendekatan garis secant secara berulang, metode ini mampu menghasilkan estimasi akar yang semakin akurat hingga mencapai batas toleransi kesalahan yang ditentukan.

```python
def secant(function: sp.Expr, x_sym: sp.Symbol, x0: float, x1: float, tolerance: float, max_iterations: int) -> tuple[float, list[tuple], int]:
    """
    Metode Secant untuk mencari akar f(x) = 0
    
    Parameters:
    - function: Fungsi yang ingin dicari akarnya, dalam bentuk ekspresi sympy
    - x_sym: Simbol variabel dalam fungsi, sebagai objek sympy.Symbol
    - x0: Tebakan pertama (initial guess 1) untuk iterasi metode Secant
    - x1: Tebakan kedua (initial guess 2) untuk iterasi metode Secant
    """
    
    # Membuat fungsi numerik dari ekspresi simbolik
    f = sp.lambdify(x_sym, function, 'numpy')
    
    # Inisialisasi variabel untuk menyimpan riwayat iterasi
    history = []
    iteration = 0
    
    # Iterasi metode Secant
    for i in range(max_iterations):
        # Hitung nilai fungsi pada dua tebakan terakhir
        # yi = f(x1) dan yi-1 = f(x0)
        yi = f(x1)
        yi_minus_1 = f(x0)
        
        # Hindari pembagian dengan nol
        if abs(yi - yi_minus_1) < 1e-12:
            print(f"Warning: Function values at {x0} and {x1} are too close. Stopping iteration.")
            break
        
        # Hitung tebakan berikutnya menggunakan rumus Secant 
        x_next = x1 - (yi * (x1 - x0) / (yi - yi_minus_1))
        
        # Hitung error sebagai selisih mutlak antara tebakan berikutnya dan tebakan terakhir
        error = abs(x_next - x1)
        
        # Simpan riwayat iterasi
        history.append((i+1, x0, yi_minus_1, x1, yi, x_next, error))  
        
        # Update nilai x0 dan x1 untuk iterasi berikutnya
        x0 = x1
        x1 = x_next
        iteration += 1
        
        # Periksa kriteria konvergensi
        if abs(error) < tolerance:
            return x_next, history, iteration

    # Jika kita mencapai batas iterasi maksimum tanpa konvergensi, kita bisa memberikan peringatan atau mengembalikan hasil terbaik yang ditemukan sejauh ini
    print("Maximum iterations reached without convergence.")
    return x1, history, iteration
```

```python
# Implementasi Kasus
# Kasus 4: Mencari akar dari fungsi f(x) = xe^-x + 1 dengan interval [-1, 0] menggunakan metode Secant
# e : euler's number
# ============================================================
METHOD = "Secant"

X_SYM = sp.Symbol('x')
FUNC = X_SYM * sp.E**(-X_SYM) + 1
X0_GUESS = -1
X1_GUESS = 0

EPSILON_TOLERANCE = 10**(-3)
MAX_ITERATIONS = 20

# ===========================================================
print_header_case(METHOD, FUNC)
# ===========================================================

# Menjalankan metode Secant
# ===========================================================
root, iteration_history, iteration_count = secant(FUNC, X_SYM, X0_GUESS, X1_GUESS, EPSILON_TOLERANCE, MAX_ITERATIONS)
print_secant_table(root, iteration_history)
# ===========================================================
```
