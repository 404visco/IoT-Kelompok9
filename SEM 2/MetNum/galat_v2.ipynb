# Deret Taylor

Dalam tugas ini, akan dilakukan perhitungan galat pemotongan (truncation error) pada aproksimasi nilai fungsi $tan(x)$ dan $\sqrt{1+x}$ pada titik **$x = 0.2$** dengan menggunakan **5 suku pertama deret taylor**. Deret Taylor merupakan metode dalam analisis numerik yang digunakan untuk mengaproksimasi nilai suatu fungsi di sekitar titik tertentu dengan memanfaatkan nilai turunan-turunan fungsi tersebut pada titik acuan.

Deret Taylor dari fungsi $f(x)$ di sekitar titik $x = a$ dapat ditulis sebagai:

$$
f(x) \approx f(a) + f'(a) \frac{(x-a)^1}{1!} + f''(a) \frac{(x-a)^2}{2!} + f'''(a) \frac{(x-a)^3}{3!} + \cdots = f(x) = \sum_{n=0}^{\infty} \frac{f^{(n)}(a)}{n!} (x-a)^n
$$

# Deret Maclaurin

Deret Maclaurin merupakan **kasus khusus dari deret Taylor**, yaitu ketika titik pengembangan diambil pada **$x = 0$**.  
Secara umum, deret Taylor mengaproksimasi suatu fungsi di sekitar titik $x = a$, sedangkan deret Maclaurin secara khusus mengaproksimasi fungsi di sekitar **titik nol**.

Deret Maclaurin dari fungsi $f(x)$ dapat dituliskan sebagai:

$$
\begin{aligned}
f(x) &\approx f(0) + f'(0)\frac{x^1}{1!} + f''(0)\frac{x^2}{2!} + f'''(0)\frac{x^3}{3!} + \cdots = f(x) = \sum_{n=0}^{\infty} \frac{f^{(n)}(0)}{n!} x^n
\end{aligned}
$$

# Fungsi Deret Maclaurin - Reusable
Implementasi algoritma Deret Maclaurin menggunakan fungsi python, ***dapat digunakan untuk kedua kasus (soal A dan soal B)**

**Soal A**

**$y = \tan(x)$**

**Metode penurunan deret:**


Akan dilakukan perbandingan nilai asli $tan(0.2)$ dalam radian dengan $tan(0.2)$ berdasarkan ekspansi deret tailor dengan orde 5 dan titik pusat yang konvergen di $(0,0)$. Pertama akan dilakukan perhitungan turunan pertama hingga ke empat dari fungsi $tan(x)$.
Kenapa hanya sampai turunan ke empat? Karena orde ke 5 dalam deret taylor cukup sampai menggunakan turunan ke empat dari fungsi.

$$
\begin{align*}
f(x) & = \tan(x) \\
f'(x) & = \sec^2(x) \\
f''(x) & = 2\sec^2(x) \tan(x) \\
f'''(x) & = 4\sec^2(x) \tan^2(x)+2 \sec^4(x)\\
f^{(4)}(x) & = 8\sec^2(x) \tan^3(x)+16 \sec^4(x) \tan(x) 
\end{align*}
$$

Kemudian, akan dilakukan evaluasi nilai fungsi pada titik pusat (0,0)

$$
\begin{align*}
f(0) & = \tan(0) = 0\\
f'(0) & = \sec^2(0) = 1\\
f''(0) & = 2\sec^2(0) \tan(0) =0\\
f'''(0) & = 4\sec^2(0) \tan^2(0)+2 \sec^4(0) = 2\\
f^{(4)}(0) & = 8\sec^2(0) \tan^3(0)+16 \sec^4(0) \tan(0) =0
\end{align*}
$$

Dengan hasil turunan tersebut, ekspansi dari aproksimasi $tan(x)$ dapat dituliskan sebagai:
$$
\tan(x) \approx f(0) + f'(0) \frac{x^1}{1!} + f''(0) \frac{x^2}{2!} + f'''(0) \frac{x^3}{3!} + f^{(4)}(0) \frac{x^4}{4!}
$$

Subtitusi nilai:
$$
\tan(x) \approx 0 + 1 \frac{x^1}{1!} + 0 \frac{x^2}{2!} + 2 \frac{x^3}{3!} + 0 \frac{x^4}{4!}
$$

Maka, ekspansi deret Taylor hingga orde 5 adalah:
$$
\tan(x) \approx x + 2\frac{x^3}{6}

Masukkan nilai $x = 0.2$ :
$$
\tan(0.2) \approx 0.2 + 2\frac{0.2^3}{6}
\tan(0.2) \approx 0.20266666667
$$

Dengan demikian, nilai aproksimasi dari $tan(0.2)$ dengan ekspansi Deret Taylor hingga orde 5 adalah $ 0.20266666667$. Nilai asli dari $tan(0.2)$ akan dihitung menggunakan numpy, yang memberikan hasil sekitar $0.20271003551$.

Galat pemotongan dapat dihitung sebagai selisih antara nilai asli dan nilai aproksimasi:

$$
\text{Galat Pemotongan} = \left| \text{Nilai Asli} - \text{Aproksimasi} \right|
$$

$$
\left| 0.20271003551 -  0.20266666667 \right| \approx 0.00004336884
$$

Dengan demikian, galat pemotongan dari aproksimasi ini adalah sekitar $0.00004336884$.

**Soal B**


**$y = \frac{1}{\sqrt{1+x}}$**

**Metode penurunan deret:**


Dengan langkah yang sama seperti sebelumnya, Akan dilakukan perbandingan nilai asli $\frac{1}{\sqrt{1+x}}$ dengan $\frac{1}{\sqrt{1+x}}$ berdasarkan ekspansi deret Taylor dengan orde 5 dan titik pusat yang konvergen di $(0,0)$. Pertama, akan dilakukan perhitungan turunan pertama hingga ke empat dari fungsi $f(x)= \frac{1}{\sqrt{1+x}}$

$$
\begin{align*}
f(x) & = \frac{1}{\sqrt{1+x}} \\
f'(x) & = -\frac{1}{2}(1+x)^{-3/2} \\
f''(x) & = \frac{3}{4}(1+x)^{-5/2} \\
f'''(x) & = -\frac{15}{8}(1+x)^{-7/2} \\
f^{(4)}(x) & = \frac{105}{16}(1+x)^{-9/2}
\end{align*}
$$

Kemudian, akan dilakukan evaluasi nilai fungsi pada titik pusat $(x = 0)$ - Deret MacLaurin:

$$
\begin{align*}
f(0) & = \frac{1}{\sqrt{1+0}}= 1\\
f'(0) & = -\frac{1}{2}(1+0)^{-3/2}= -\frac{1}{2}\\
f''(0) & = \frac{3}{4}(1+0)^{-5/2}= \frac{3}{4}\\
f'''(0) & = -\frac{15}{8}(1+0)^{-7/2}= -\frac{15}{8}\\
f^{(4)}(0) & = \frac{105}{16}(1+0)^{-9/2}= \frac{105}{16}
\end{align*}
$$



Dengan hasil turunan tersebut, ekspansi dari aproksimasi $\frac{1}{\sqrt{1+x}}$ dapat dituliskan sebagai:
$$
\frac{1}{\sqrt{1+x}} \approx f(0) + f'(0) \frac{x^1}{1!} + f''(0) \frac{x^2}{2!} + f'''(0) \frac{x^3}{3!} + f^{(4)}(0) \frac{x^4}{4!}
$$

Subtitusi:
$$
\frac{1}{\sqrt{1+x}} \approx 1 - \frac{1 \times x^1}{2 \times 1!} + \frac{3 \times x^2}{4 \times 2!} - \frac{15 \times x^3}{8 \times 3!} + \frac{105 \times x^4}{16 \times 4!}
$$

Maka Ekspansi Deret Maclaurin hingga orde 5 adalah:
$$
\begin{aligned}
\frac{1}{\sqrt{1+x}} \approx 1 - \frac{x^1}{2}  + \frac{3x^2}{8} - \frac{15x^3}{48}  +\frac{105x^4}{384}
\end{aligned}
$$

Masukkan nilai $x = 0.2$ :
$$
\begin{aligned}
\frac{1}{\sqrt{1+0.2}} \approx 1 - \frac{0.2^1}{2}  + \frac{3 \times 0.2^2}{8} - \frac{15 \times 0.2^3}{48}  +\frac{105 \times 0.2^4}{384}\\
\frac{1}{\sqrt{1.2}} \approx 1 - 0.1  + 0.015- 0.0025 + 0.0004375\\
\frac{1}{\sqrt{1.2}} \approx 0.9129375
\end{aligned}
$$

Dengan demikian, nilai aproksimasi dari $\frac{1}{\sqrt{1+0.2}}$ dengan ekspansi Deret MacLaurin hingga orde 5 adalah $0.9129375$. Nilai asli dari $\frac{1}{\sqrt{1+0.2}}$ akan dihitung menggunakan numpy, yang memberikan hasil sekitar $0.9128709291752769$.

Galat pemotongan dapat dihitung sebagai selisih antara nilai asli dan nilai aproksimasi:

$$
\text{Galat Pemotongan} = \left| \text{Nilai Asli} - \text{Aproksimasi} \right|
$$

$$
\left| 0.9128709291752769 -  0.9129375 \right| \approx 0.00006657082472305564
$$

Dengan demikian, galat pemotongan dari aproksimasi ini adalah sekitar $0.00006657082472305564$.

# Program

```python
# Import library yang dibutuhkan
import numpy as np
import sympy as sp
from tabulate import tabulate

#nilai x dan n sesuai ketentuan tugas
x_value= 0.2
n = 5
```

# Fungsi deret Maclaurin
Implementasi algoritma Deret Maclaurin dengan fungsi python - **Reusable (Bisa diterapkan untuk soal A dan B)**

```python
#Fungsi Factorial
def factorial(n):
  if n == 0:
    return 1
  else:
    return n * factorial(n-1)

#Fungsi Deret MacLaurin
def deret_maclaurin(fungsi,nilai_asli, x_val, n):
  x = sp.Symbol('x') #Buat variabel x agar fungsi bisa disubtitusi 
  fx = fungsi
  hasil= 0
  for i in range(n):
    turunan = sp.diff(fx,x,i) #sp.diff(ekspresi, variabel, orde) --> untuk hitung nilai diferensial
    turunan_x_0= turunan.subs(x,0) #subtitusi x =0
    hasil+= turunan_x_0 * x_val**i /factorial(i)
    galat = abs(nilai_asli-hasil)
    print(f'Galat Orde {i+1}: {galat}') #Untuk memberi tahu nilai galat di tiap orde
  return hasil
```

**Soal A**

**$y = \tan(x)$**

**Pemanggilan program untuk soal A:**

```python
#Deklarasi Fungsi
x = sp.Symbol('x') #mendefinisikan x sebagai variabel simbolik (bisa disubtitusi)
fungsi = sp.tan(x)

#nilai asli dihitung dengan numpy
tan_exact= np.tan(0.2)

#nilai tan yang dicari dengan deret maclaurin
tan_approx= deret_maclaurin(fungsi,tan_exact,x_value,n)

#Galat Absolut
galat_absolut= abs(tan_exact - tan_approx)

#Galat Relatif
galat_relatif= galat_absolut/tan_exact
print(80*'=')
print(f'Nilai tan(0.2) asli: {tan_exact}')
print(f'Nilai Aproksimasi tan(0.2) dengan Deret Taylor dan MacLaurin: {tan_approx}')
print(f'Nilai galat absolut: {galat_absolut}')
print(f'Nilai galat relatif: {galat_relatif}')
```

```python
#Visualisasi Tabel
data = [
    ["Method", "Value"],
    ["Nilai tan asli", tan_exact],
    ["Nilai tan approximasi", tan_approx],
    ["Galat Absolut", galat_absolut],
    ["Galat Relatif", galat_relatif],
]

print(tabulate(data, headers="firstrow", tablefmt="grid", floatfmt=".20f"))
```

**Soal B**

**$y = \frac{1}{\sqrt{1+x}}$**

**Pemanggilan program untuk soal A:**

```python
#Deklarasi Fungsi
x= sp.Symbol('x') #mendefinisikan x sebagai variabel simbolik (bisa disubtitusi)
fungsi = 1/sp.sqrt(1+x)

#nilai asli dihitung dengan numpy
y_exact= 1/np.sqrt(1.2)

#nilai y yang dicari dengan deret mac laurin
y_approx= deret_maclaurin(fungsi, y_exact, x_value, n) #Pemanggilan Fungsi

#Galat Absolut
galat_absolut= abs(y_exact - y_approx)

#Galat Relatif
galat_relatif= galat_absolut/y_exact

print(75*'=')
print(f'Nilai y asli: {y_exact}')
print(f'Nilai Aproksimasi y dengan Deret Taylor dan MacLaurin: {y_approx}')
print(f'Nilai galat absolut: {galat_absolut}')
print(f'Nilai galat relatif: {galat_relatif}')
```

```python
#Visualisasi Tabel
data = [
    ["Method", "Value"],
    ["Nilai tan asli", y_exact],
    ["Nilai tan approximasi", y_approx],
    ["Galat Absolut", galat_absolut],
    ["Galat Relatif", galat_relatif],
]

print(tabulate(data, headers="firstrow", tablefmt="grid", floatfmt=".20f"))
```
