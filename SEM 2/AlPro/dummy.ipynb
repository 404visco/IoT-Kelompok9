```python
import random
D = random.randint(0,100)
K = random.randint(0,100)
M = random.randint(0,100)
print(f'Durian: {D}, Kesemek: {K}, Melon: {M}')
Db = 0
Kb = 0
Mb = 0
Sd= None
Sk= None
Sm= None
while D>=3 or K>=3 or M>=3:
    if D>=3:
        Db+=1
        D-=3
    if K>=3:
        Kb+=1
        K-=3
    if M>=3:
        Mb+=1
        M-=3
    else:
        Kb+= 0
        Mb+= 0
        Db+= 0
Sd= D
Sk= K
Sm= M
print(f'Durian Dibagikan: {Db}, Sisa: {Sd}')
print(f'Kesemek Dibagikan: {Kb}, Sisa: {Sk}')
print(f'Melon Dibagikan: {Mb}, Sisa: {Sm}')
```
