### **Properti #1:**

Menambah atau mengurangi dengan suatu nilai konstanta untuk semua dataset. Untuk menemukan properti pertama, kita akan mencari nilai mean (rata-rata) dan median (nilai tengah) dari kumpulan data, lalu menambahkan (atau mengurangi) suatu nilai konstanta yang sama untuk setiap baris dalam data, dan menemukan nilai rata-rata dan median yang baru.

**Langkah 1:**

Hitung nilai standard deviation dan variance sebelum memodifikasi dataset untuk
variabel Tipe3

```python
df.head()
```

```python
#SEBELUM

print('=== Sebelum ===')
std_before = df['Tipe3'].std()
var_before = df['Tipe3'].var()
print(f'Standar deviasi Tipe3   : {std_before}')
print(f'Variansi Tipe3          : {var_before}')
```

**Langkah 2:**

Modifikasi satu buah variabel dari dataset terkait dengan menambahkan atau
mengurangi suatu konstanta untuk setiap data.

```python
#SETELAH ditambah Konstanta
df['Tipe3'] -= 8
df.head()
```

**Langkah 3:**

Hitung nilai standard deviation dan variance dari modified dataset untuk melengkapi tabel pada Langkah 1 dan lakukan analisa terhadap hasil yang diperoleh sesuai pertanyaan berikut ini.

```python
print('=== Setelah ===')
std_after = df['Tipe3'].std()
var_after = df['Tipe3'].var()
print(f'Standar Deviasi Tipe3   : {std_after}')
print(f'Variansi Tipe3          : {var_after}')
```

Setelah dilakukan penambahan konstanta pada seluruh data, diperoleh bahwa nilai standar deviasi dan variansi tidak mengalami perubahan. Hal ini terjadi karena penambahan konstanta hanya menggeser seluruh data tanpa mengubah jarak relatif antar data terhadap nilai rata-rata.

Dengan demikian, dapat disimpulkan bahwa variansi dan standar deviasi bersifat invariant (tidak berubah) terhadap penjumlahan atau pengurangan konstanta.

### **Properti #2:**

Mengalikan atau membagi dengan suatu nilai konstanta untuk semua dataset.

**Langkah 1:**

Hitung nilai standard deviation dan variance sebelum memodifikasi dataset untuk
variabel Tipe2

```python
df.head()
```

```python
#SEBELUM
print('=== Sebelum ===')
std_before = df['Tipe2'].std()
var_before = df['Tipe2'].var()
print(f'Standar Deviasi Tipe2   : {std_before}')
print(f'Variasi Tipe2           : {var_before}')
```

**Langkah 2:**

Modifikasi satu buah variabel dari dataset terkait dengan menambahkan suatu
konstanta untuk setiap data.

```python
#SETELAH ditambah Konstanta
df['Tipe2'] *= -4

df.head()
```

**Langkah 3:**

Hitung nilai standard deviation dan variance dari modified dataset untuk
melengkapi tabel pada Langkah 1 dan lakukan analisa terhadap hasil yang diperoleh sesuai
pertanyaan berikut ini.

```python
print('=== Setelah ===')
std_after = df['Tipe2'].std()
var_after = df['Tipe2'].var()
print(f'Standar Deviasi Tipe2   : {std_after}')
print(f'Variansi Tipe2          : {var_after}')
```

Setelah seluruh data dikalikan dengan suatu konstanta, nilai variansi berubah menjadi sebesar kuadrat dari konstanta tersebut dikalikan dengan variansi awal, sedangkan standar deviasi berubah menjadi sebesar nilai absolut konstanta dikalikan dengan standar deviasi awal.

Dengan demikian, dapat disimpulkan bahwa variansi bersifat kuadratik terhadap konstanta, sedangkan standar deviasi bersifat linear terhadap konstanta.

**Langkah 3:**

Nilai yang jauh dari mean menghasilkan selisih yang besar, dan pada variansi selisih tersebut dikuadratkan sehingga efeknya semakin besar. Akibatnya, standar deviasi juga ikut meningkat. Dengan demikian, variansi dan standar deviasi sangat sensitif terhadap outlier.
